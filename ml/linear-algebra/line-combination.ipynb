```python
def scalar(X: list, alpha: float):
    return list(map(lambda x: x * alpha, X))

def scalar_multiple(X: list[list], w: list, k: int) -> list:
    new_matrix = []

    for scale_i in range(k):
        new_matrix.append(scalar(X[scale_i], w[scale_i]))

    new_vector = []

    for el in list(zip(*new_matrix)):
        new_vector.append(sum(el))

    return new_vector

n = int(input())
w = list(map(float, input().split()))

matrix = []

for i in range(n):
    matrix.append(list(map(int, input().split())))

print(*scalar_multiple(matrix, w, n))
```

```python
import sys


def main():
    """
    Пример ввода и вывода числа n, где -10^9 < n < 10^9:
    n = int(input())
    print(n)
    """
    
    def scalar(X: list, alpha: float):
        return list(map(lambda x: x * alpha, X))

    def scalar_multiple(X: list[list], w: list, k: int) -> list:
        new_matrix = []

        for scale_i in range(k):
            new_matrix.append(scalar(X[scale_i], w[scale_i]))

        new_vector = []

        for el in list(zip(*new_matrix)):
            new_vector.append(sum(el))

        return new_vector

    n = int(input())
    w = list(map(float, input().split()))

    matrix = []

    for _ in range(n):
        matrix.append(list(map(int, input().split())))

    print(*scalar_multiple(matrix, w, n))

if __name__ == '__main__':
    main()
```
