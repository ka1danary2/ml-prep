## Лекция 3 Классы

Класс - это способ организации кода

Мы группируем данные и связанные с ними методы для того, чтобы было удобнее работать

При создании инстанста происходит следующее:


1) Метод `__new__` создает объект и выделяет память
2) `__init__` инициализирует уже созданный объект, устанавливает его атрибуты


`self` - ссылка на конкретный экзепляр класса


У класса есть 2 вида атрибутов:
- атрибуты экземпляра - принадлежат конкретному инстансу
- атрибуты класса - принадлежат всему классу и всем его инстансам

Атрибут класса общий для всех экземлпятров, которые не переопределелили его у себя

При присваивании значения через экземпляр (obj.attr = value) обычно создаётся или изменяется атрибут самого экземпляра. Если в классе существует одноимённый атрибут, он не изменяется — атрибут экземпляра просто перекрывает его при поиске

Если атрибут - это изменяемый объект, то его можно менять через атрибут экзмпляра, так как это изменяемый объект

`super()` - это встроенная функция python, которая позволяет обращаться к методам и атрибутам родительского класса в порядке наследования

```python
class Dog:
    a = 1

    def __init__(self, name: str):
        self.name = name

dog = Dog("Dog 1")

print(dog.a)  # 1

dog.a = 54

print(dog.a)  # 54
print(Dog.a)  # 1
```

При обращении к атрибуту поиск происходит в следующем порядке:
- Сначала смотрим на экземпляр
- Потом на класс
- Потом на родительские классы

Если атрибута нет, то возвращается `AttributeError`

Название `self` - это просто соглашение, на самом деле на месте первого аргумента может стоять все что угодно

`self` - это ссылка на текущий экземпляр класса, для которого вызвается метод

```python
class Dog:

    # функция присвоения полям инстанса данных, обязана возвращать ничего
    def __init__(self, name: str) -> None:
        self.name  = name    
        self.ball = None

    # это обычный метод класса
    def bark(self) -> None:
        print('Woof')    

    # это обычный метод для изменения состояния собаки
    def catch_ball(self, ball: str):
        if self.ball is not None:
            self.ball = None
            self.bark()
            return

        self.ball = ball

    # тоже обычный метод
    def take_ball(self) -> str:
        ball = self.ball
        self.ball = None

        return ball

    # магичесике методы

    # str() - человекочитаемое представление объекта, обычно используется print()
    def __str__(self) -> str:
        return f'Dog name is {self.name}'

    # вернуть строку которая предсталяет инстанс (объект) (reprezentation)
    def __repr__(self):
        return f'Dog name is from __repr {repr(self.name)}'

    # отвечает за сложение, когда инстанс слева (dog + 'Мяч')
    def __add__(self, ball: str):
        self.catch_ball(ball)
        # результатом будет сам инстанс изменнного dog, а не новый результат операции
        return self

    # отвечает за отраженное сложение, когда объект находится справа ('Мяч' + dog) (reflected add)
    def __radd__(self, ball: str):
            self.catch_ball(ball)
            # результатом будет сам инстанс изменнного dog, а не новый результат операции
            return self

    # // - определение метода деления с округлением
    def __floordiv__(self, value: int) ->list[str]:
        return [Dog(name=self.name) for i in range(value)]

    # / - определение методов деления с остатком
    def __truediv__(self, value: int) -> list["Dog"]:
        return [Dog(name=self.name) for i in range(value)]

    # метод сравнения инстансов через оператор ==
    def __eq__(self, value):
        return self.name == value.name

    # определяет операцию len() для инстанса класса
    def __len__(self):
        return len(self.name)

    # определяет операцию взятия элемента по индексу
    def __getitem__(self, index):
        return self.name[index]

    # вызывается при обычном обращении к атрибуту экземпляра
    def __getattribute__(self, name):
        print('Ищем атрибут')
        # передача поиска стандартной операции взятия атрибута для стандартной реализации родительского класса
        return super().__getattribute__(name)

    # вызывается когда поиск атрибута через метод __getattribute__ вернулся с ошибкой AttributeError
    def __getattr__(self, name):
        return f'какой то текс {name}'

    # перехват во время присвоения значний
    def __setattr__(self, name, value):
        print('Перехватчик присвоения определен')
        super().__setattr__(name, value)

    # оперделяет список возвращаемых имен для функции dir()
    # помогает узнать какие методы и атрибуты доступны у инстанса
    #  
    def __dir__(self):
        print('Функция dir переопределена')
        return super().__dir__()

    ## срабатывает в момент финализации инстанса, то есть в момент удаления его из памяти в процессе сборки мусора
    # но на него не стоит полагаться, так как сборка мусора в питоне происходит не сразу, а по расписанию и зависит от многих факторов и от циклических зависимостей, поэтому ничего кртически важного тут делать нельзя
    def __del__(self):
        print('dog died')

    def __iter__(self):
        pass

    def __next__(self):
        pass

    def __(self, name):
        print(name)
        return super().__getattribute__(name)
```

## NotImplementedError

Бызово в питоне нет абстрактных методов

В питоне нет проверки реализации методов и на этапе написания кода, мы не можем понять, реализован ли родительский метод ребенком

Чтобы это как-то косо обойти, можно прописывать методы с raise NotImplementedError, чтобы при вызове метода через ребенка мы понимали, что забыла реализовать его. Но это сработает только в момент вызова

Для возможности использования абстрактных методов есть декораты из модуля ABC

```python
from abc import ABC, abstractmethod
```

`class Room(ABC):` - в параметрах класса надо указывать родительские методы, от которых мы наследуемся

ABC - abstract base

```python
import random

class Room(ABC):

    def __init__(self, number: int, building: str, capacity: int):
        self.number = number
        self.building = building
        self.capacity = capacity
        self.is_occupid = False

    def str(self) -> str:
        return f'{self.building}{self.number}'

    def occupy(self):
        assert not self.is__occupied, 'ЗАНЯТО GO AWAY'

    def free(self):
        assert self.is_occupid, 'DOUBLE FREE OR CORRUPTION'

    # в питоне нет проверки реализации методов и на этапе написания кода, мы не можем понять, реализован ли родительский метод ребенком

    # чтобы это как-то косо обойти, можно прописывать методы с raise NotImplementedError, чтобы при вызове метода через ребенка мы понимали, что забыла реализовать его. Но это сработает только в момент вызова

    def look(self):
        raise NotImplementedError

    @abstractmethod
    def sing(self):
        print('singing')
```

```python
# super() - вывзает метод от родителя у которого он есть

# передаем родительсикй класс в параметр
class Classroom(Room):
    def __init__(
        self,
        number: int,
        building: str,
        capacity: int,
        enable_multimedia: bool
    ):
        super().__init__(number, building, capacity)
        self.enable_multimedia = enable_multimedia

    def enable_multimedias(self):
        self.enable_multimedia = random.random() < 0.5

    def sing(self):
        print('Реализация абстрактного метода')

    def disble_multimedia(self):
         self.enable_multimedia = False
         
    def look(self):
        print('CANNOT LOCK')
```

```python
room = Classroom(101, "A", 30, True)
```

```python
# наследование
class Office(Room):
    def __init__(self, number: int, building: str, capacity: int, *employees: str):
        super().__init__(number, building, capacity)
        self.employees = list(employees)

        assert len(self.employees) <= capacity

    def fire(self, employees: str):
        self.employees.remove(employees)

    def sing(self):
        print('Реализация абстрактного метода') 


    def __str__(self):
        return f'{super().__str__()}({" ".join(self.employees)})'
    
```

изменить тип в питоне нельзя потому что типов и кастов вообще нет в питоне

```python
office = Office(123, 'A', 2, ['Masha', 'Pasha', 'Tolia'])
```

# Датаклассы

## @property

Это свойства класса, которые позволяют управлять чтением, изменением, и удалением атрибутов

Это специальный дескриптор, который позволяет образаться к методам как к атрибутам объекта

Дексриптор в питоне - это атрибут класса, который переоределяет стандартное поведение при чтении, записи или удалении через специальные методы

- `__get__()`
- `__set__()`
- `__delete__()`

Представим обычный класс

```python
class Student:
    def __init__(self, age: int, name: str):
        self.age = age
        self.name = name

    def get_age(self):
        return self.age
```

Для получения возраста мы должны вызывать метод

```python
student = Student(23, 'Max')

student.get_age()
# или так student.age
```

Теперь попробуем использовать `property`

```python
class Student:
    def __init__(self, age: int, name: str):
        self._age = age
        self.name = name

    @property
    def age(self) -> int:
        print("Вызвался геттер")
        return self._age


student = Student(23, "Max")

print(student.age)
```

Как можно заметить, это просто синтаксический сахар
Пользователю кажется, что он просто получает возраст, но на самом деле это вычисляемое поле

В целом можно использовать просто функцию без property и разницы не будет, это все еще будет вычисляемое значение

Рассмотрим на более репризентативном примере

Мы хотим площадь квадрата возвращать

```python
from dataclasses import dataclass

@dataclass
class Rectangle:
    width: int
    height: int

    @property
    def area(self):
        return self.width * self.height


rect = Rectangle(10, 2)
rect.area
```

То есть пользователю нашего кода будет казаться, что площадь это поле, хотя на самом деле оно вычисляется через метод на лету

В целом можно было сделать и так:

```python
from dataclasses import dataclass

@dataclass
class Rectangle:
    width: int
    height: int

    def get_area(self):
        return self.width * self.height


rect = Rectangle(10, 2)
rect.get_area()
```

## setter

Нужен для попытки присваивания нашему `property` атрибуту значения

`property` выглядит для пользователя как обычный атрибут, значит кто-то явно попытается к нему присвоить значение

Для этого существет `@property.setter `, который позволяет перехватить присваивание
