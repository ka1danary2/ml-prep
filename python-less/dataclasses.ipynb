```python
from dataclasses import dataclass, field
```

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # вывод осмысленного выражения в консоль
    def __repr__(self):
        return f'Экземпляр класса Dog с именем {self.name} и возрастом {self.age}'

dog1 = Dog('Бобик', 3)
dog1
```

декоратор dataclass предоставляет более удобный интерфейс для реализации класса, например не нужно писать init

```python
@dataclass
class DogData:
    name: str
    age: int

dog2 = DogData('Bobik', 4)
dog2
```

Также у dataclass есть встроенный метод `__eg__` который сравнивает инстансы не по id, а по полям и сравнивает их

(field_1, field_2, ...) = (field_1, field_2, ...)

```python
dog2 = DogData('Bobik', 4)
dog3 = DogData('Bobik', 4)

dog2 == dog3
```

Как и любой магический метод его можно переопределить внутри класса

Порядок позиционных и непозиционных аргументов важен в dataclass

```python
@dataclass
class DogData:
    name: str
    age: int = 0
    color: str


dog4 = DogData('Bobik', 4)
```

Очевидно делать так нельзя потому что у аргументов в питоне должен быть порядок, а этот декоратор это надстройка над `__init__`, где аргументы передаются в скобки в определенном порядке

В параметры по дефолту можно передавать спокойно лишь неизменяемые значения

Иначе будут известные баги с тем, что параметр ссылается на один и тот же объект а не создает новый

```python

@dataclass
class DogData:
    name: str
    age: int
    color: list = []


dog4 = DogData('Bobik', 4)
dog4.color.append('black')
```

Ну тут даже ошибка из компилятора вылетела, надо использовать функцию `field` которая создает уникальный список дефолтный

```python

@dataclass
class DogData:
    name: str
    age: int
    color: list = field(default_factory=list)


dog4 = DogData('Bobik', 4)
dog4.color.append('black')
dog5 = DogData('Bobik', 5)
dog5.color.append('black')
(dog4.color, dog5.color)
```

Получили независимые дефолтные массивы

```python
[1, 2] == [1, 2]
```
